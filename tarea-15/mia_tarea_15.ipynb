# 3. Cargamos el modelo y lo optimizamos

```python
!pip install -U diffusers accelerate transformers torch huggingface_hub
```

```python
import torch
from diffusers import FluxPipeline

pipe = FluxPipeline.from_pretrained(
    "black-forest-labs/FLUX.1-schnell",
    torch_dtype=torch.bfloat16
)

pipe.enable_sequential_cpu_offload()
pipe.enable_vae_slicing()
pipe.enable_vae_tiling()

generator = torch.Generator(device="cpu").manual_seed(0)
```

# 4. Generamos la imagen

```python
prompt = "A futuristic smart city with flying cars, ultra realistic, cinematic lighting"

image = pipe(
    prompt,
    guidance_scale=0.0,
    num_inference_steps=4,
    max_sequence_length=256,
    generator=generator
).images[0]

image.save("resultado.png")
```

## 6.1 Analista de sentimientos

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

result = classifier("I love artificial intelligence!")
print(result)
```

## 6.2 Reconocimiento de entidades

```python
from transformers import pipeline

ner = pipeline(
    "ner",
    model="mrm8488/bert-spanish-cased-finetuned-ner",
    aggregation_strategy="simple"
)

texto = "Pedro Sánchez vive en Madrid y trabaja para el Gobierno de España."

print(ner(texto))
```

## 6.3 Contestador de respuestas


```python
!pip install --upgrade transformers
```

```python
from transformers import pipeline

qa = pipeline("text-generation", model="deepset/roberta-base-squad2")

context = "Carlos is a software developer who works in Madrid."
question = "Where does Carlos work?"

prompt = f"Context: {context}\nQuestion: {question}\nAnswer:"
result = qa(prompt, max_new_tokens=50)
print(result[0]["generated_text"])
```

## 6.4 Traductor

```python
from transformers import MarianMTModel, MarianTokenizer

model_name = "Helsinki-NLP/opus-mt-en-fr"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

text = "Artificial intelligence is amazing"
inputs = tokenizer(text, return_tensors="pt", padding=True)
translated = model.generate(**inputs)
result = tokenizer.decode(translated[0], skip_special_tokens=True)
print(result)
```

## 6.5 De texto a voz

```python
from transformers import pipeline
import soundfile as sf

tts = pipeline("text-to-speech", model="facebook/mms-tts-spa")

audio = tts("La inteligencia artificial está transformando el mundo.")

sf.write("voz.wav", audio["audio"], samplerate=audio["sampling_rate"])
```
